```python
import numpy as np

PI2 = 2 * np.pi
signal = [0.8, -0.3, 0.4, -0.55]
spectrum = np.fft.fft(signal)
print("Спектр (numpy.fft):", spectrum)
```

```python
def compute_dft(samples):
    n = len(samples)
    time_points = np.linspace(0, 1, n, endpoint=False)
    freq_bins = np.arange(n)
    phase_matrix = np.outer(time_points, freq_bins)
    transform_matrix = np.exp(1j * PI2 * phase_matrix)
    return transform_matrix.conj().T @ samples

dft_result = compute_dft(signal)
print("Спектр (DFT):", dft_result)
print("Разница:", np.sum(np.abs(spectrum - dft_result)))
```

```python
def simple_fft(samples):
    n = len(samples)
    even_fft = np.fft.fft(samples[::2])
    odd_fft = np.fft.fft(samples[1::2])
    
    k = np.arange(n)
    twiddle = np.exp(-1j * PI2 * k / n)
    
    return np.tile(even_fft, 2) + twiddle * np.tile(odd_fft, 2)

fft_non_recursive = simple_fft(signal)
print("Спектр (FFT без рекурсии):", fft_non_recursive)
print("Разница:", np.sum(np.abs(spectrum - fft_non_recursive)))
```

```python
def recursive_fft(samples):
    n = len(samples)
    if n == 1:
        return samples
    
    even_part = recursive_fft(samples[::2])
    odd_part = recursive_fft(samples[1::2])
    
    k = np.arange(n)
    twiddle = np.exp(-1j * PI2 * k / n)
    
    return np.tile(even_part, 2) + twiddle * np.tile(odd_part, 2)
fft_recursive = recursive_fft(signal)
print("Спектр (FFT с рекурсией):", fft_recursive)
print("Разница:", np.sum(np.abs(spectrum - fft_recursive)))
```

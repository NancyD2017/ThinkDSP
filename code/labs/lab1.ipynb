```python
from thinkdsp import read_wave
from thinkdsp import decorate

wave = read_wave('760961__audiocoffee__inspiring-dream-short-ver.wav')
wave.normalize()
wave.make_audio()
```

```python
wave.plot()
decorate(xlabel='Время (с)')
```

```python
segment = wave.segment(start=7.1, duration=0.5)
segment.make_audio()
```

```python
segment.plot()
decorate(xlabel='Время (с)')
```

```python
spectrum = segment.make_spectrum()
spectrum.plot()
decorate(xlabel='Частота (Гц)')
```

```python
spectrum.plot(high=4000)
decorate(xlabel='Частота (Гц)')
```

```python
spectrum.low_pass(500)
spectrum.plot(high=500)
decorate(xlabel='Частота (Гц)')
```

```python
spectrum.make_wave().make_audio()
```

```python
filtered = spectrum.make_wave()
filtered.normalize()
filtered.plot()
decorate(xlabel='Время (с)')
```

```python
spectrum.make_wave().make_audio()
```

Упражнение 1.3

```python
from thinkdsp import CosSignal, SinSignal
from thinkdsp import decorate

cos_sig = CosSignal(freq=80, amp=0.75)
sin_sig = SinSignal(freq=205, amp=1.0)

cos_sig.plot()
decorate(xlabel='Время (с)')

sin_sig.plot()
decorate(xlabel='Время (с)')
```

```python
signal = cos_sig + sin_sig
signal.plot()
decorate(xlabel='Время (с)')
```

```python
firstWave = signal.make_wave(duration=1.5, start=0, framerate=11025)
firstWave.make_audio()
```

```python
spectrum = firstWave.make_spectrum()
spectrum.plot(high=550)
decorate(xlabel='Частота (Гц)')
```

```python
signal += SinSignal(freq=30)
signal.make_wave().make_audio()
```

```python
secondWave = signal.make_wave()
secondWave.apodize
spectrum2 = secondWave.make_spectrum()
spectrum2.plot(high=550)
decorate(xlabel='Частота (Гц)')
```

```python
def stretch(wave, factor):
    wave.ts *= factor
    wave.framerate /= factor
```

```python
stretch(wave, 2.0)
wave.make_audio()
```

```python
wave.plot()
decorate(xlabel='Время (с)')
```
